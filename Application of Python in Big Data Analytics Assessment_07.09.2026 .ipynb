### 1. Write a program to accept for an integer input and print a number pattern in rows where each row contains a number repeated the same number of times as its value. E.g The following number pattern is for an integer 5. [5]


```python
#accepting input from the user
number = int(input('Enter an integer : '))

for i in range(1, number + 1):
    for j in range(i):
        print(i, end = ' ')
    print('\n')
```

### 2. Write a Python function that takes the following list of numbers and returns the third largest value. 
####    [13, 30, 2, 11, 15, 2, 30, 15]. 
### Ensure the function handles lists with duplicate values correctly e.g., if the list is [12, 12, 5, 6], the third largest is 5. [10]

```python
#function decleration
def third_largest_val(the_list):
    unique_numbers = list(set(the_list))
    unique_numbers.sort()
    return unique_numbers[2]

#initialising the list
the_list = [13, 30, 2, 11, 15, 2, 30, 15]
print(third_largest_val(the_list))
```

### 3. Write a program to create a list of 20 integers, count the occurrences of each integer, and create a dictionary to store and show the count of each element. [10]

```python
#import random library
import random

#list to store the 20 integers
integers = [38, 1, 7, 38, 5, 7, 38, 7, 45, 39, 32, 61, 89, 89, 50, 50, 30, 69, 38, 61]

#dictionary to store the occurances of each element
val_count = {}

#checking the number of occurances for each element int the list
for i in integers:
    if i in val_count:
        val_count[i] += 1
    else:
        val_count[i] = 1
print('Original List : ', integers)
print('Count of elements : ', val_count)
```

### 4. Write a programm to capture the id, name, surname, age and gender of 5 people and create a dataframe. Export the dataframe to an Excel file without the index. [10]

```python
#handling empty cells int the 'focus_score' column
focus_score = sp_df['focus_score'].median()
sp_df['focus_score'] = sp_df['focus_score'].fillna(focus_score)
```

```python
#handling empty cells int the 'mental_state' column
mental_state = sp_df['mental_state'].mode()[0]
sp_df['mental_state'] = sp_df['mental_state'].fillna(mental_state)
```

```python
#handling empty cells int the 'sleep_hours' column
sleep_hours = sp_df['sleep_hours'].median()
sp_df['sleep_hours'] = sp_df['sleep_hours'].fillna(sleep_hours)
```

```python
#handling empty cells int the 'physical_activity_hours' column
physical_activity_hours = sp_df['physical_activity_hours'].median()
sp_df['physical_activity_hours'] = sp_df['physical_activity_hours'].fillna(physical_activity_hours)
```

```python
#handling empty cells int the 'learning_style' column
learning_style = sp_df['learning_style'].mode()[0]
sp_df['learning_style'] = sp_df['learning_style'].fillna(learning_style)
```

```python
#handling empty cells int the 'productivity_after_midnight' column
productivity_after_midnight = sp_df['productivity_after_midnight'].median()
sp_df['productivity_after_midnight'] = sp_df['productivity_after_midnight'].fillna(productivity_after_midnight)
```

```python
#handling empty cells int the 'burnout_risk' column
burnout_risk = sp_df['burnout_risk'].median()
sp_df['burnout_risk'] = sp_df['burnout_risk'].fillna(burnout_risk)
```

```python
#handling empty cells int the 'consistency_score' column
consistency_score = sp_df['consistency_score'].median
sp_df['burnout_risk'] = sp_df['consistency_score'].fillna(consistency_score)
```

```python
#encoding the non-numeric columns using One-Hot Encoding
encoded_sp_df = pd.get_dummies(sp_df, columns = ['gender', 'city_type', 'mental_state', 'learning_style', 'career_goal', 'performance_category'])
```

```python
#exporting/saving the clean dataset to excel
encoded_sp_df.to_excel('Encoded Student Performance.xlsx', index = False)
```

### 9. Build a predictive model to estimate medical insurance charges based on demographic, lifestyle, and geographic details. Training and testing sets should be split on a ratio of 80:20. Evaluate the performance of the model using an appropriate evaluation metric. [15]

```python
#reading/loading the dataset
insurance = pd.read_csv('Insurance.csv')

#standardising the columns
insurance.columns = insurance.columns.str.strip().str.lower().str.replace(' ', '_')
insurance.head()
```

```python
#encoding 'sex' column using Cat.codes method
insurance['encoded_gender'] = insurance['sex'].astype('category').cat.codes
insurance.head()
```

```python
#encoding 'smoking_status' column using Cat.codes method
insurance['encoded_smoking_status'] = insurance['smoking_status'].astype('category').cat.codes
insurance.head()
```

```python
insurance['encoded_region'] = insurance['region'].astype('category').cat.codes
insurance
```

```python
#defining the features (X and Y variables)
x = insurance.drop(columns = ['sex', 'smoking_status', 'region', 'charges'])
y = insurance[['charges']]
```

```python
#splitting the data into training and testing sets
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2, random_state = 42)
```

```python
#building the linear regression model 
from sklearn.linear_model import LinearRegression
lm = LinearRegression()
lm
```

```python
#training the model
lm.fit(x_train, y_train)
lm
```

```python
#testing the model using the test data
predicted_insurance = lm.predict(x_test)
predicted_insurance = pd.DataFrame(predicted_insurance)
```

```python
from sklearn.metrics import r2_score
import numpy as np
r2_score = r2_score(y_test, predicted_insurance)
print(r2_score)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sb

# Scatter plot
plt.figure(figsize=(8,6))
plt.scatter(y_test, predicted_insurance, alpha=0.6, edgecolors='k', linewidth=0.3)
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], 'r--', label='Insurance Prediction')
plt.xlabel('Actual Charges')
plt.ylabel('Predicted Charges')
plt.title('Actual vs Predicted Insurance Charges')
plt.legend()
plt.grid(True, alpha=0.3)
plt.savefig('actual_vs_predicted.png', dpi=300)
plt.show()
```
